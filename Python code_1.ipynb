```python
import pandas as pd
import numpy as np

np.random.seed(42)

reference_year = 2026

def generate_dob(dept):

    if dept == "DEP001":
        age = np.random.randint(22, 59)

    elif dept == "DEP002":
        age = np.random.randint(24, 61)

    elif dept == "DEP003":
        age = np.random.randint(23, 59)

    elif dept == "DEP004":
        age = np.random.randint(24, 59)

    elif dept == "DEP005":
        age = np.random.randint(22, 56)

    else:
        age = np.random.randint(23, 59)

    year = reference_year - age

    month = np.random.randint(1, 13)
    day = np.random.randint(1, 29)

    return pd.Timestamp(year=year, month=month, day=day)

dim_employee["Date_of_Birth"] = dim_employee["Department_ID"].apply(generate_dob)
```

```python
# ==========================================
# 12. HIRE DATE
# ==========================================

import numpy as np
import pandas as pd

np.random.seed(42)

service_groups = np.random.choice(
    [
        "0-2",
        "3-5",
        "6-10",
        "11-20",
        "21-30"
    ],
    size=num_employees,
    p=[
        0.20,
        0.25,
        0.30,
        0.20,
        0.05
    ]
)

hire_dates = []

for group in service_groups:

    if group == "0-2":
        year = np.random.randint(2024, 2027)

    elif group == "3-5":
        year = np.random.randint(2021, 2024)

    elif group == "6-10":
        year = np.random.randint(2016, 2021)

    elif group == "11-20":
        year = np.random.randint(2006, 2016)

    else:
        year = np.random.randint(1995, 2006)

    month = np.random.randint(1, 13)
    day = np.random.randint(1, 28)

    hire_dates.append(pd.Timestamp(year, month, day))

dim_employee["Hire_Date"] = hire_dates

dim_employee["Hire_Date"].head()
```

```python
# ==========================================
# 15. SUPERVISOR ID
# ==========================================

import numpy as np

np.random.seed(42)

# Default: no supervisor assigned yet
dim_employee["Supervisor_ID"] = None

for dept in dim_employee["Department_ID"].unique():

    dept_index = dim_employee[
        dim_employee["Department_ID"] == dept
    ].index

    dept_employees = dim_employee.loc[dept_index, "Employee_ID"].values

    # About 12% become supervisors
    n_supervisors = max(1, int(len(dept_employees) * 0.12))

    supervisors = np.random.choice(
        dept_employees,
        size=n_supervisors,
        replace=False
    )

    for idx in dept_index:

        employee = dim_employee.loc[idx, "Employee_ID"]

        # Supervisors report to themselves (top of team)
        if employee in supervisors:
            dim_employee.loc[idx, "Supervisor_ID"] = employee

        else:
            dim_employee.loc[idx, "Supervisor_ID"] = np.random.choice(
                supervisors
            )

dim_employee[
    ["Employee_ID", "Department_ID", "Supervisor_ID"]
].head(20)
```

```python
incident_type_weights = {
    "IT001": 0.50,   # Near Miss
    "IT002": 0.16,   # First Aid Case
    "IT003": 0.10,   # Medical Treatment Case
    "IT004": 0.06,   # Restricted Work Case
    "IT005": 0.035,  # Lost Time Injury
    "IT006": 0.002,  # Fatality
    "IT007": 0.06,   # Property Damage
    "IT008": 0.025,  # Environmental Incident
    "IT009": 0.025,  # Fire / Explosion
    "IT010": 0.033   # Process Safety Event
}

print("Total Weight:", sum(incident_type_weights.values()))
```

```python
# ============================================================
# STEP 19 — INCIDENT TYPE PROFILE
# ============================================================

def get_incident_type_weights(department_id, asset_id):

    weights = incident_type_weights.copy()

    # --------------------------------------------------------
    # HIGH-HAZARD OPERATING ASSETS
    # Offshore, gas, refinery, LNG
    # --------------------------------------------------------

    high_hazard_assets = [
        "AST001", "AST002", "AST003",  # Offshore
        "AST006", "AST007",              # Gas
        "AST008",                        # Refinery
        "AST010"                         # LNG
    ]

    if asset_id in high_hazard_assets:
        weights["IT009"] *= 1.8   # Fire / Explosion
        weights["IT010"] *= 1.8   # Process Safety Event
        weights["IT008"] *= 1.2   # Environmental Incident

    # --------------------------------------------------------
    # MAINTENANCE
    # --------------------------------------------------------

    if department_id == "DEP003":
        weights["IT002"] *= 1.25   # First Aid
        weights["IT003"] *= 1.30   # MTC
        weights["IT004"] *= 1.20   # Restricted Work
        weights["IT005"] *= 1.15   # LTI
        weights["IT007"] *= 1.20   # Property Damage

    # --------------------------------------------------------
    # OPERATIONS
    # --------------------------------------------------------

    if department_id == "DEP001":
        weights["IT001"] *= 1.15   # Near Miss
        weights["IT009"] *= 1.25
        weights["IT010"] *= 1.30

    # --------------------------------------------------------
    # LOGISTICS
    # --------------------------------------------------------

    if department_id == "DEP005":
        weights["IT007"] *= 1.30   # Property Damage
        weights["IT002"] *= 1.15   # First Aid

    # --------------------------------------------------------
    # OFFICE-BASED FUNCTIONS
    # --------------------------------------------------------

    if department_id in ["DEP006", "DEP007"]:
        weights["IT001"] *= 1.30   # Near Miss
        weights["IT002"] *= 1.15   # First Aid
        weights["IT009"] *= 0.30
        weights["IT010"] *= 0.25
        weights["IT008"] *= 0.40

    # --------------------------------------------------------
    # NORMALIZE BACK TO 1
    # --------------------------------------------------------

    total = sum(weights.values())

    weights = {
        k: v / total
        for k, v in weights.items()
    }

    return weights
```

```python
# ============================================================
# STEP 24 — HAZARD PROFILE
# ============================================================

def get_hazard_weights(department_id, asset_id):

    weights = {
        "HZ001": 0.14,  # Slips, Trips & Falls
        "HZ002": 0.08,  # Working at Height
        "HZ003": 0.04,  # Confined Space
        "HZ004": 0.08,  # Electrical Energy
        "HZ005": 0.09,  # Pressure / Stored Energy
        "HZ006": 0.07,  # Fire / Explosion
        "HZ007": 0.07,  # Hazardous Chemicals
        "HZ008": 0.10,  # Hydrocarbon Release
        "HZ009": 0.10,  # Lifting & Material Handling
        "HZ010": 0.08,  # Vehicle / Mobile Equipment
        "HZ011": 0.07,  # Dropped Objects
        "HZ012": 0.08   # Ergonomic / Manual Handling
    }

    # --------------------------------------------------------
    # HIGH-HAZARD PROCESS ASSETS
    # --------------------------------------------------------

    high_hazard_assets = [
        "AST001", "AST002", "AST003",
        "AST006", "AST007",
        "AST008", "AST010"
    ]

    if asset_id in high_hazard_assets:
        weights["HZ005"] *= 1.5
        weights["HZ006"] *= 1.5
        weights["HZ008"] *= 1.7
        weights["HZ002"] *= 1.3
        weights["HZ011"] *= 1.3

    # --------------------------------------------------------
    # OPERATIONS
    # --------------------------------------------------------

    if department_id == "DEP001":
        weights["HZ008"] *= 1.4
        weights["HZ005"] *= 1.3
        weights["HZ006"] *= 1.2
        weights["HZ011"] *= 1.2

    # --------------------------------------------------------
    # MAINTENANCE
    # --------------------------------------------------------

    if department_id == "DEP003":
        weights["HZ004"] *= 1.5
        weights["HZ005"] *= 1.4
        weights["HZ009"] *= 1.4
        weights["HZ001"] *= 1.2

    # --------------------------------------------------------
    # LOGISTICS
    # --------------------------------------------------------

    if department_id == "DEP005":
        weights["HZ010"] *= 1.6
        weights["HZ009"] *= 1.5
        weights["HZ001"] *= 1.2

    # --------------------------------------------------------
    # OFFICE FUNCTIONS
    # --------------------------------------------------------

    if department_id in ["DEP006", "DEP007"]:
        weights["HZ012"] *= 1.8
        weights["HZ001"] *= 1.3
        weights["HZ004"] *= 0.5
        weights["HZ005"] *= 0.4
        weights["HZ006"] *= 0.3
        weights["HZ008"] *= 0.3
        weights["HZ011"] *= 0.4

    # --------------------------------------------------------
    # NORMALIZE
    # --------------------------------------------------------

    total = sum(weights.values())

    weights = {
        k: v / total
        for k, v in weights.items()
    }

    return weights
```

```python
# ============================================================
# STEP 28 — CONTRACTOR INVOLVEMENT PROBABILITY
# ============================================================

def get_contractor_probability(department_id, asset_id):

    probability = 0.15

    # Higher contractor presence in technical/field departments
    department_probability = {
        "DEP001": 0.25,  # Operations
        "DEP002": 0.30,  # Engineering
        "DEP003": 0.40,  # Maintenance
        "DEP004": 0.15,  # HSE
        "DEP005": 0.35,  # Logistics
        "DEP006": 0.05,  # Finance
        "DEP007": 0.05   # HR
    }

    probability = department_probability.get(
        department_id,
        probability
    )

    # Higher contractor presence around major operating assets
    high_contractor_assets = [
        "AST001", "AST002", "AST003",  # Offshore
        "AST004", "AST005",             # Pipeline
        "AST006", "AST007",             # Gas
        "AST008",                       # Refinery
        "AST010"                        # LNG
    ]

    if asset_id in high_contractor_assets:
        probability *= 1.20

    # Keep probability within sensible bounds
    probability = min(probability, 0.60)

    return probability
```

```python
# ============================================================
# STEP 34 — DOWNTIME HOURS
# ============================================================

np.random.seed(42)

def generate_downtime(incident_type):

    if incident_type == "IT001":       # Near Miss
        return np.random.choice(
            [0, np.random.uniform(0.5, 4)],
            p=[0.90, 0.10]
        )

    elif incident_type == "IT002":     # First Aid
        return np.random.choice(
            [0, np.random.uniform(0.5, 8)],
            p=[0.80, 0.20]
        )

    elif incident_type == "IT003":     # Medical Treatment
        return np.random.uniform(0, 12)

    elif incident_type == "IT004":     # Restricted Work
        return np.random.uniform(1, 24)

    elif incident_type == "IT005":     # LTI
        return np.random.uniform(4, 48)

    elif incident_type == "IT006":     # Fatality
        return np.random.uniform(24, 168)

    elif incident_type == "IT007":     # Property Damage
        return np.random.uniform(2, 72)

    elif incident_type == "IT008":     # Environmental
        return np.random.uniform(1, 48)

    elif incident_type == "IT009":     # Fire / Explosion
        return np.random.uniform(12, 240)

    elif incident_type == "IT010":     # Process Safety Event
        return np.random.uniform(8, 168)

    return 0


fact_hse_incident["Downtime_Hours"] = (
    fact_hse_incident["Incident_Type_ID"]
    .apply(generate_downtime)
    .round(2)
)
```

```python
# ============================================================
# STEP 35 — CALIBRATED HSE FINANCIAL IMPACT
# ============================================================

np.random.seed(42)

def generate_hse_costs(row):

    incident_type = row["Incident_Type_ID"]
    downtime = row["Downtime_Hours"]
    lost_days = row["Lost_Days"]

    # --------------------------------------------------------
    # 1. PROPERTY DAMAGE
    # --------------------------------------------------------

    if incident_type == "IT001":          # Near Miss
        property_cost = np.random.uniform(0, 100)

    elif incident_type == "IT002":        # First Aid
        property_cost = np.random.uniform(0, 500)

    elif incident_type == "IT003":        # Medical Treatment
        property_cost = np.random.uniform(0, 1_000)

    elif incident_type == "IT004":        # Restricted Work
        property_cost = np.random.uniform(250, 2_500)

    elif incident_type == "IT005":        # LTI
        property_cost = np.random.uniform(500, 5_000)

    elif incident_type == "IT006":        # Fatality
        property_cost = np.random.uniform(1_000, 15_000)

    elif incident_type == "IT007":        # Property Damage
        property_cost = np.random.uniform(5_000, 75_000)

    elif incident_type == "IT008":        # Environmental
        property_cost = np.random.uniform(500, 10_000)

    elif incident_type == "IT009":        # Fire / Explosion
        property_cost = np.random.uniform(10_000, 100_000)

    elif incident_type == "IT010":        # Process Safety
        property_cost = np.random.uniform(5_000, 75_000)

    # --------------------------------------------------------
    # 2. ENVIRONMENTAL REMEDIATION
    # --------------------------------------------------------

    if incident_type == "IT008":
        environmental_cost = np.random.uniform(5_000, 100_000)

    elif incident_type == "IT009":
        environmental_cost = np.random.uniform(1_000, 30_000)

    elif incident_type == "IT010":
        environmental_cost = np.random.uniform(1_000, 25_000)

    else:
        environmental_cost = np.random.uniform(0, 1_000)

    # --------------------------------------------------------
    # 3. EMERGENCY RESPONSE
    # --------------------------------------------------------

    if incident_type == "IT001":
        emergency_cost = np.random.uniform(0, 100)

    elif incident_type == "IT002":
        emergency_cost = np.random.uniform(100, 750)

    elif incident_type == "IT003":
        emergency_cost = np.random.uniform(250, 2_000)

    elif incident_type == "IT004":
        emergency_cost = np.random.uniform(500, 3_000)

    elif incident_type == "IT005":
        emergency_cost = np.random.uniform(1_000, 7_500)

    elif incident_type == "IT006":
        emergency_cost = np.random.uniform(5_000, 30_000)

    elif incident_type == "IT007":
        emergency_cost = np.random.uniform(1_000, 15_000)

    elif incident_type == "IT008":
        emergency_cost = np.random.uniform(1_000, 20_000)

    elif incident_type == "IT009":
        emergency_cost = np.random.uniform(10_000, 75_000)

    elif incident_type == "IT010":
        emergency_cost = np.random.uniform(5_000, 50_000)

    # --------------------------------------------------------
    # 4. MEDICAL / COMPENSATION
    # --------------------------------------------------------

    if incident_type == "IT001":
        medical_cost = 0

    elif incident_type == "IT002":
        medical_cost = np.random.uniform(100, 1_500)

    elif incident_type == "IT003":
        medical_cost = np.random.uniform(1_000, 10_000)

    elif incident_type == "IT004":
        medical_cost = np.random.uniform(2_000, 15_000)

    elif incident_type == "IT005":
        medical_cost = np.random.uniform(5_000, 50_000)

    elif incident_type == "IT006":
        medical_cost = np.random.uniform(50_000, 200_000)

    else:
        medical_cost = 0

    # --------------------------------------------------------
    # 5. OPERATIONAL DISRUPTION
    # Conservative rate so HSE remains below pipeline costs
    # --------------------------------------------------------

    disruption_rate = np.random.uniform(100, 500)

    operational_cost = (
        downtime * disruption_rate
        + lost_days * np.random.uniform(100, 500)
    )

    # --------------------------------------------------------
    # 6. OTHER HSE COSTS
    # --------------------------------------------------------

    other_cost = np.random.uniform(0, 2_000)

    # --------------------------------------------------------
    # TOTAL HSE INCIDENT COST
    # --------------------------------------------------------

    total_cost = (
        property_cost
        + environmental_cost
        + emergency_cost
        + medical_cost
        + operational_cost
        + other_cost
    )

    return (
        property_cost,
        environmental_cost,
        emergency_cost,
        medical_cost,
        operational_cost,
        other_cost,
        total_cost
    )


# Generate costs
hse_costs = fact_hse_incident.apply(
    generate_hse_costs,
    axis=1
)

# Add columns
fact_hse_incident["Property_Damage_Cost"] = (
    hse_costs.apply(lambda x: round(x[0], 2))
)

fact_hse_incident["Environmental_Cost"] = (
    hse_costs.apply(lambda x: round(x[1], 2))
)

fact_hse_incident["Emergency_Response_Cost"] = (
    hse_costs.apply(lambda x: round(x[2], 2))
)

fact_hse_incident["Medical_Compensation_Cost"] = (
    hse_costs.apply(lambda x: round(x[3], 2))
)

fact_hse_incident["Operational_Disruption_Cost"] = (
    hse_costs.apply(lambda x: round(x[4], 2))
)

fact_hse_incident["Other_Cost"] = (
    hse_costs.apply(lambda x: round(x[5], 2))
)

fact_hse_incident["Total_Incident_Cost"] = (
    hse_costs.apply(lambda x: round(x[6], 2))
)
```

```python
# ============================================================
# STEP 4 — ACTION TIMELINE & CLOSURE STATUS
# REPORTING CUTOFF: JULY 31, 2026
# ============================================================

np.random.seed(42)

reference_date = pd.Timestamp("2026-07-31")

# ------------------------------------------------------------
# 1. Make sure Incident_Date is datetime
# ------------------------------------------------------------

fact_hse_corrective_action["Incident_Date"] = pd.to_datetime(
    fact_hse_corrective_action["Incident_Date"]
)

# ------------------------------------------------------------
# 2. Action Date
# ------------------------------------------------------------
# Action occurs 1–14 days after the incident,
# but cannot go beyond the reporting cutoff.

days_available = (
    reference_date
    - fact_hse_corrective_action["Incident_Date"]
).dt.days

action_delay = np.array([
    np.random.randint(1, min(14, days_available_i) + 1)
    if days_available_i >= 1
    else 0
    for days_available_i in days_available
])

fact_hse_corrective_action["Action_Date"] = (
    fact_hse_corrective_action["Incident_Date"]
    + pd.to_timedelta(action_delay, unit="D")
)

# ------------------------------------------------------------
# 3. Target Closure Days
# ------------------------------------------------------------

def closure_target(priority):

    if priority == "Critical":
        return np.random.randint(7, 31)

    elif priority == "High":
        return np.random.randint(14, 46)

    elif priority == "Medium":
        return np.random.randint(30, 76)

    else:
        return np.random.randint(45, 121)


fact_hse_corrective_action["Target_Closure_Days"] = (
    fact_hse_corrective_action["Priority"]
    .apply(closure_target)
)

# ------------------------------------------------------------
# 4. Due Date
# ------------------------------------------------------------
# Due dates can extend beyond July 31 because they represent
# the deadline assigned to an action.

fact_hse_corrective_action["Due_Date"] = (
    fact_hse_corrective_action["Action_Date"]
    + pd.to_timedelta(
        fact_hse_corrective_action["Target_Closure_Days"],
        unit="D"
    )
)

# ------------------------------------------------------------
# 5. Action Status
# ------------------------------------------------------------

fact_hse_corrective_action["Action_Status"] = np.random.choice(
    [
        "Closed",
        "Open",
        "Overdue"
    ],
    size=len(fact_hse_corrective_action),
    p=[
        0.68,
        0.20,
        0.12
    ]
)

# ------------------------------------------------------------
# 6. Closure Date
# ------------------------------------------------------------
# Closed actions must have closed between Action_Date
# and July 31, 2026.
#
# If an action is too recent to allow a 5-day closure period,
# it will not be assigned a Closed status.

days_available_for_closure = (
    reference_date
    - fact_hse_corrective_action["Action_Date"]
).dt.days

closed_mask = (
    fact_hse_corrective_action["Action_Status"] == "Closed"
)

# Identify closed actions that have at least 5 days available
valid_closed_mask = (
    closed_mask
    & (days_available_for_closure >= 5)
)

# Closed actions without enough time to have closed
# are converted to Open.
fact_hse_corrective_action.loc[
    closed_mask & ~valid_closed_mask,
    "Action_Status"
] = "Open"

# Recalculate closed mask after adjustment
closed_mask = (
    fact_hse_corrective_action["Action_Status"] == "Closed"
)

days_available_for_closure = (
    reference_date
    - fact_hse_corrective_action["Action_Date"]
).dt.days

# Start with no closure dates
fact_hse_corrective_action["Closure_Date"] = pd.NaT

# Generate closure dates only for valid closed actions
for idx in fact_hse_corrective_action.index[closed_mask]:

    max_days = int(
        days_available_for_closure.loc[idx]
    )

    closure_days = np.random.randint(
        5,
        max_days + 1
    )

    fact_hse_corrective_action.loc[
        idx,
        "Closure_Date"
    ] = (
        fact_hse_corrective_action.loc[
            idx,
            "Action_Date"
        ]
        + pd.Timedelta(days=closure_days)
    )

# ------------------------------------------------------------
# 7. Days Open
# ------------------------------------------------------------

fact_hse_corrective_action["Days_Open"] = np.where(

    fact_hse_corrective_action["Closure_Date"].notna(),

    (
        fact_hse_corrective_action["Closure_Date"]
        - fact_hse_corrective_action["Action_Date"]
    ).dt.days,

    (
        reference_date
        - fact_hse_corrective_action["Action_Date"]
    ).dt.days
)

# ------------------------------------------------------------
# 8. Overdue Flag
# ------------------------------------------------------------

fact_hse_corrective_action["Overdue_Flag"] = np.where(
    fact_hse_corrective_action["Action_Status"] == "Overdue",
    1,
    0
)

# ------------------------------------------------------------
# 9. Closed Flag
# ------------------------------------------------------------

fact_hse_corrective_action["Closed_Flag"] = np.where(
    fact_hse_corrective_action["Action_Status"] == "Closed",
    1,
    0
)

# ============================================================
# STEP 4 CHECK
# ============================================================

print("========== STEP 4 CHECK ==========")

print("\nAction Status:")
print(
    fact_hse_corrective_action[
        "Action_Status"
    ].value_counts()
)

print("\nMinimum Action Date:",
      fact_hse_corrective_action["Action_Date"].min())

print("Maximum Action Date:",
      fact_hse_corrective_action["Action_Date"].max())

print("\nMinimum Due Date:",
      fact_hse_corrective_action["Due_Date"].min())

print("Maximum Due Date:",
      fact_hse_corrective_action["Due_Date"].max())

print("\nMinimum Closure Date:",
      fact_hse_corrective_action["Closure_Date"].min())

print("Maximum Closure Date:",
      fact_hse_corrective_action["Closure_Date"].max())

print("\nMissing Action Dates:",
      fact_hse_corrective_action["Action_Date"].isna().sum())

print("Missing Due Dates:",
      fact_hse_corrective_action["Due_Date"].isna().sum())

print("Missing Closure Dates:",
      fact_hse_corrective_action["Closure_Date"].isna().sum())

print("\nNegative Days Open:",
      (
          fact_hse_corrective_action["Days_Open"] < 0
      ).sum())

print("Minimum Days Open:",
      fact_hse_corrective_action["Days_Open"].min())

print("Maximum Days Open:",
      fact_hse_corrective_action["Days_Open"].max())

print("\nAction Dates After Reporting Cutoff:",
      (
          fact_hse_corrective_action["Action_Date"]
          > reference_date
      ).sum())

print("\nClosure Dates After Reporting Cutoff:",
      (
          fact_hse_corrective_action["Closure_Date"]
          > reference_date
      ).sum())

print("\nOverdue Flag:")
print(
    fact_hse_corrective_action[
        "Overdue_Flag"
    ].value_counts()
)

print("\nClosed Flag:")
print(
    fact_hse_corrective_action[
        "Closed_Flag"
    ].value_counts()
)
```

```python
# ============================================================
# STEP 19 — REVISED INCIDENT TYPE PROFILE
# REALISTIC HSE SEVERITY DISTRIBUTION
# ============================================================

incident_type_weights = {
    "IT001": 0.58,   # Near Miss
    "IT002": 0.18,   # First Aid Case
    "IT003": 0.07,   # Medical Treatment Case
    "IT004": 0.035,  # Restricted Work Case
    "IT005": 0.015,  # Lost Time Injury
    "IT006": 0.001,  # Fatality
    "IT007": 0.055,  # Property Damage
    "IT008": 0.020,  # Environmental Incident
    "IT009": 0.018,  # Fire / Explosion
    "IT010": 0.026,  # Process Safety Event
}

print("Total Weight:", sum(incident_type_weights.values()))
```

```python
# ============================================================
# STEP 19B — HSE INCIDENT TYPE PROFILE BY RISK CONTEXT
# ============================================================

def get_incident_type_weights(department_id, asset_id):

    weights = incident_type_weights.copy()

    # --------------------------------------------------------
    # HIGH-HAZARD OPERATING ASSETS
    # Offshore, gas, refinery, LNG
    # --------------------------------------------------------

    high_hazard_assets = [
        "AST001", "AST002", "AST003",
        "AST006", "AST007",
        "AST008",
        "AST010"
    ]

    if asset_id in high_hazard_assets:
        weights["IT009"] *= 1.40   # Fire / Explosion
        weights["IT010"] *= 1.40   # Process Safety
        weights["IT008"] *= 1.15   # Environmental

    # --------------------------------------------------------
    # MAINTENANCE
    # --------------------------------------------------------

    if department_id == "DEP003":
        weights["IT002"] *= 1.10   # First Aid
        weights["IT003"] *= 1.15   # Medical Treatment
        weights["IT004"] *= 1.10   # Restricted Work
        weights["IT005"] *= 1.05   # LTI
        weights["IT007"] *= 1.15   # Property Damage

    # --------------------------------------------------------
    # OPERATIONS
    # --------------------------------------------------------

    if department_id == "DEP001":
        weights["IT001"] *= 1.10   # Near Miss
        weights["IT009"] *= 1.15
        weights["IT010"] *= 1.20

    # --------------------------------------------------------
    # LOGISTICS
    # --------------------------------------------------------

    if department_id == "DEP005":
        weights["IT007"] *= 1.20
        weights["IT002"] *= 1.10

    # --------------------------------------------------------
    # OFFICE-BASED FUNCTIONS
    # --------------------------------------------------------

    if department_id in ["DEP006", "DEP007"]:
        weights["IT001"] *= 1.20
        weights["IT002"] *= 1.10
        weights["IT009"] *= 0.40
        weights["IT010"] *= 0.35
        weights["IT008"] *= 0.50

    # --------------------------------------------------------
    # NORMALIZE
    # --------------------------------------------------------

    total = sum(weights.values())

    weights = {
        k: v / total
        for k, v in weights.items()
    }

    return weights
```
